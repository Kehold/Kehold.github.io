```python
# Comprehensive Exploratory Data Analysis for Sabah and Abu Dhabi Reservoir Characterization
# Final fixed version

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import ndimage
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.feature_selection import mutual_info_regression, mutual_info_classif
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)

class ReservoirEDA:
    def __init__(self):
        self.sabah_data = None
        self.abudhabi_data = None
        self.sabah_features = None
        self.abudhabi_features = None
        
    def generate_realistic_sabah_data(self, nt=50, nx=50, ny=50):
        """Generate realistic synthetic seismic and well data for Sabah turbidite systems"""
        print("Generating realistic Sabah data...")
        
        # Create seismic data with channel features
        seismic_data = np.random.normal(0, 0.3, (nt, nx, ny))
        
        # Add channel features using vectorized operations
        x_coords, y_coords = np.meshgrid(np.arange(nx), np.arange(ny), indexing='ij')
        
        for i in range(nt):
            channel_center_x = nx // 2 + int(10 * np.sin(2 * np.pi * i / nt))
            channel_center_y = ny // 2 + int(8 * np.cos(2 * np.pi * i / nt))
            
            # Vectorized distance calculation
            distance = np.sqrt((x_coords - channel_center_x)**2 + (y_coords - channel_center_y)**2)
            
            # Vectorized channel creation with realistic geometry
            channel_width = 8 + 3 * np.sin(0.2 * x_coords) + 2 * np.random.random()
            channel_mask = distance < channel_width
            seismic_data[i, channel_mask] += 0.7 * np.exp(-distance[channel_mask]/8)
            
            # Add some random noise to simulate real seismic uncertainty
            seismic_data[i, :, :] += 0.2 * np.random.normal(0, 0.1, (nx, ny))
        
        # Add some layered structure with more variability
        for i in range(nt):
            layer_intensity = 0.3 * np.sin(0.05 * i) + 0.5 * np.random.random()
            seismic_data[i, :, :] += layer_intensity * np.sin(0.1 * i + 0.5 * np.random.random())
        
        # Apply smoothing
        seismic_data = ndimage.gaussian_filter(seismic_data, sigma=1.2)
        
        # Generate well data with realistic relationships
        n_wells = 5
        wells = {}
        
        well_positions = [
            (nx//4, ny//4), (nx//4, 3*ny//4), 
            (nx//2, ny//2), (3*nx//4, ny//4), 
            (3*nx//4, 3*ny//4)
        ]
        
        for i, (x, y) in enumerate(well_positions):
            seismic_trace = seismic_data[:, x, y]
            
            # Create sand/shale classification with realistic, noisy relationship
            sand_probability = (seismic_trace - np.min(seismic_trace)) / (np.max(seismic_trace) - np.min(seismic_trace))
            sand_probability = np.clip(sand_probability + 0.2 * np.random.random(nt) - 0.1, 0, 1)
            
            # Use probability to determine facies with some misclassifications
            facies = (sand_probability > (0.5 + 0.2 * np.random.random())).astype(int)
            random_mask = np.random.random(nt) < 0.15  # 15% misclassification
            facies[random_mask] = 1 - facies[random_mask]
            
            # Add missing values to simulate real data issues (5% missing)
            missing_mask = np.random.random(nt) < 0.05
            seismic_trace_with_nan = seismic_trace.copy()
            seismic_trace_with_nan[missing_mask] = np.nan
            
            wells[f'Well_{i+1}'] = {
                'position': (x, y),
                'seismic_trace': seismic_trace_with_nan,
                'facies': facies,
                'permeability': 5 + 150 * sand_probability + 30 * np.random.random(nt),
                'porosity': 0.05 + 0.25 * sand_probability + 0.08 * np.random.random(nt)
            }
        
        self.sabah_data = {
            'seismic': seismic_data,
            'wells': wells,
            'dimensions': (nt, nx, ny)
        }
        
        return self.sabah_data
    
    def generate_realistic_abudhabi_data(self, nt=50, nx=50, ny=50):
        """Generate realistic synthetic seismic and well data for Abu Dhabi carbonate systems"""
        print("Generating realistic Abu Dhabi data...")
        
        # Create seismic data with layered carbonate features
        seismic_data = np.random.normal(0, 0.3, (nt, nx, ny))
        
        # Add layered structure with diagenetic features
        sweet_spot_probability = np.zeros((nt, nx, ny))
        
        for i in range(nt):
            # Create horizontal layers with some gentle folding and variability
            layer_intensity = 0.3 * np.sin(0.05 * i) + 0.4 * np.random.random() + 0.3
            seismic_data[i, :, :] += layer_intensity * np.sin(0.1 * i + 0.3 * np.random.random())
            
            # Create sweet spots with realistic distribution
            for x in range(0, nx, 8):
                for y in range(0, ny, 8):
                    if np.random.random() < 0.2:  # 20% chance of sweet spot nucleation
                        size = 4 + 3 * np.random.random()
                        intensity = 0.5 + 0.5 * np.random.random()
                        
                        # Create a window around the sweet spot
                        x_min, x_max = max(0, int(x-size)), min(nx, int(x+size))
                        y_min, y_max = max(0, int(y-size)), min(ny, int(y+size))
                        
                        # Create distance grid
                        xx, yy = np.meshgrid(np.arange(x_min, x_max), np.arange(y_min, y_max), indexing='ij')
                        distance = np.sqrt((xx - x)**2 + (yy - y)**2)
                        
                        # Apply sweet spot effect with decay
                        mask = distance < size
                        sweet_spot_probability[i, x_min:x_max, y_min:y_max][mask] = np.maximum(
                            sweet_spot_probability[i, x_min:x_max, y_min:y_max][mask],
                            intensity * np.exp(-distance[mask]/size)
                        )
            
            # Add sweet spot effect to seismic
            seismic_data[i, :, :] += 0.6 * sweet_spot_probability[i, :, :]
            
            # Add some random noise
            seismic_data[i, :, :] += 0.2 * np.random.normal(0, 0.1, (nx, ny))
        
        # Apply smoothing
        seismic_data = ndimage.gaussian_filter(seismic_data, sigma=1.2)
        
        # Generate well data with realistic relationships
        n_wells = 5
        wells = {}
        
        well_positions = [
            (nx//4, ny//4), (nx//4, 3*ny//4), 
            (nx//2, ny//2), (3*nx//4, ny//4), 
            (3*nx//4, 3*ny//4)
        ]
        
        for i, (x, y) in enumerate(well_positions):
            seismic_trace = seismic_data[:, x, y]
            sweet_prob_trace = sweet_spot_probability[:, x, y]
            
            # Create permeability and porosity values with realistic relationships
            permeability = np.ones(nt) * (5 + 10 * np.random.random())
            porosity = np.ones(nt) * (0.1 + 0.05 * np.random.random())
            
            # Sweet spots have higher permeability and porosity
            sweet_spots = sweet_prob_trace > (0.3 + 0.2 * np.random.random())
            permeability[sweet_spots] = 50 + 150 * sweet_prob_trace[sweet_spots] + 50 * np.random.random(np.sum(sweet_spots))
            porosity[sweet_spots] = 0.15 + 0.2 * sweet_prob_trace[sweet_spots] + 0.1 * np.random.random(np.sum(sweet_spots))
            
            # Add some medium quality zones with more randomness
            medium_zones = (sweet_prob_trace > 0.1) & ~sweet_spots
            permeability[medium_zones] = 20 + 30 * sweet_prob_trace[medium_zones] + 15 * np.random.random(np.sum(medium_zones))
            porosity[medium_zones] = 0.12 + 0.1 * sweet_prob_trace[medium_zones] + 0.05 * np.random.random(np.sum(medium_zones))
            
            # Add some random noise to all values
            permeability += 10 * np.random.random(nt)
            porosity += 0.03 * np.random.random(nt)
            
            # Add missing values to simulate real data issues (5% missing)
            missing_mask = np.random.random(nt) < 0.05
            seismic_trace_with_nan = seismic_trace.copy()
            seismic_trace_with_nan[missing_mask] = np.nan
            permeability_with_nan = permeability.copy()
            permeability_with_nan[missing_mask] = np.nan
            porosity_with_nan = porosity.copy()
            porosity_with_nan[missing_mask] = np.nan
            
            # Define sweet spots with some uncertainty
            sweet_spot = (permeability > (40 + 20 * np.random.random())).astype(int)
            random_mask = np.random.random(nt) < 0.15  # 15% misclassification
            sweet_spot[random_mask] = 1 - sweet_spot[random_mask]
            
            wells[f'Well_{i+1}'] = {
                'position': (x, y),
                'seismic_trace': seismic_trace_with_nan,
                'permeability': permeability_with_nan,
                'porosity': porosity_with_nan,
                'sweet_spot': sweet_spot
            }
        
        self.abudhabi_data = {
            'seismic': seismic_data,
            'wells': wells,
            'dimensions': (nt, nx, ny)
        }
        
        return self.abudhabi_data
    
    def extract_features_from_wells(self, dataset_type):
        """Extract features from well data for EDA"""
        if dataset_type == 'sabah':
            data = self.sabah_data
            target_columns = ['facies', 'permeability', 'porosity']
        else:
            data = self.abudhabi_data
            target_columns = ['sweet_spot', 'permeability', 'porosity']
        
        # Extract features from all wells
        features_list = []
        
        for well_name, well_data in data['wells'].items():
            # Create basic features from seismic trace
            seismic_trace = well_data['seismic_trace']
            
            # Calculate various statistical features
            well_features = {
                'well': well_name,
                'mean_amplitude': np.nanmean(seismic_trace),
                'std_amplitude': np.nanstd(seismic_trace),
                'max_amplitude': np.nanmax(seismic_trace),
                'min_amplitude': np.nanmin(seismic_trace),
                'range_amplitude': np.nanmax(seismic_trace) - np.nanmin(seismic_trace),
                'rms_amplitude': np.sqrt(np.nanmean(seismic_trace**2)),
                'abs_mean_amplitude': np.nanmean(np.abs(seismic_trace)),
                'q10_amplitude': np.nanpercentile(seismic_trace, 10),
                'q90_amplitude': np.nanpercentile(seismic_trace, 90),
            }
            
            # Add skewness and kurtosis with NaN handling
            try:
                well_features['skewness_amplitude'] = pd.Series(seismic_trace).skew()
            except:
                well_features['skewness_amplitude'] = np.nan
                
            try:
                well_features['kurtosis_amplitude'] = pd.Series(seismic_trace).kurtosis()
            except:
                well_features['kurtosis_amplitude'] = np.nan
            
            # Add target variables
            for target in target_columns:
                if target in ['permeability', 'porosity']:
                    well_features[target] = np.nanmean(well_data[target])
                else:
                    # For categorical targets, use the most common value
                    # Handle NaN values by filtering them out
                    target_values = well_data[target]
                    valid_values = target_values[~np.isnan(target_values)] if target_values.dtype == np.float64 else target_values
                    
                    if len(valid_values) > 0:
                        # Use pandas to get the mode
                        mode_result = pd.Series(valid_values).mode()
                        if len(mode_result) > 0:
                            well_features[target] = mode_result.iloc[0]
                        else:
                            well_features[target] = np.nan
                    else:
                        well_features[target] = np.nan
            
            features_list.append(well_features)
        
        # Convert to DataFrame
        features_df = pd.DataFrame(features_list)
        
        if dataset_type == 'sabah':
            self.sabah_features = features_df
        else:
            self.abudhabi_features = features_df
        
        return features_df
    
    def handle_missing_values(self, df, strategy='mean'):
        """Handle missing values in the dataset"""
        # Check for missing values
        missing = df.isnull().sum()
        if missing.sum() == 0:
            print("No missing values found.")
            return df
        
        print(f"Missing values before imputation:\n{missing[missing > 0]}")
        
        # Separate numeric and categorical columns
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        categorical_cols = df.select_dtypes(exclude=[np.number]).columns
        
        # Impute numeric columns
        if strategy == 'mean':
            imputer = SimpleImputer(strategy='mean')
        elif strategy == 'median':
            imputer = SimpleImputer(strategy='median')
        elif strategy == 'knn':
            imputer = KNNImputer(n_neighbors=3)
        else:
            imputer = SimpleImputer(strategy='constant', fill_value=0)
        
        df_imputed = df.copy()
        if len(numeric_cols) > 0:
            df_imputed[numeric_cols] = imputer.fit_transform(df[numeric_cols])
        
        # For categorical columns, use mode imputation
        if len(categorical_cols) > 0:
            for col in categorical_cols:
                mode_val = df[col].mode()[0] if len(df[col].mode()) > 0 else 'Unknown'
                df_imputed[col].fillna(mode_val, inplace=True)
        
        print("Missing values handled successfully.")
        return df_imputed
    
    def perform_univariate_analysis(self, df, dataset_name):
        """Perform univariate analysis on the dataset"""
        print(f"\n=== Univariate Analysis for {dataset_name} ===")
        
        # Separate numeric and categorical features
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        categorical_cols = df.select_dtypes(exclude=[np.number]).columns
        
        # Plot distributions for numeric features
        n_cols = 3
        n_rows = int(np.ceil(len(numeric_cols) / n_cols))
        
        fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5*n_rows))
        axes = axes.flatten()
        
        for i, col in enumerate(numeric_cols):
            if i < len(axes):
                df[col].hist(bins=30, ax=axes[i])
                axes[i].set_title(f'Distribution of {col}')
                axes[i].set_xlabel(col)
                axes[i].set_ylabel('Frequency')
        
        # Hide any empty subplots
        for j in range(i+1, len(axes)):
            fig.delaxes(axes[j])
        
        plt.tight_layout()
        plt.suptitle(f'Univariate Analysis - {dataset_name}', y=1.02, fontsize=16)
        plt.show()
        
        # Print summary statistics
        print("\nSummary Statistics:")
        print(df[numeric_cols].describe())
        
        # For categorical features, show value counts
        for col in categorical_cols:
            print(f"\nValue counts for {col}:")
            print(df[col].value_counts())
    
    def perform_bivariate_analysis(self, df, dataset_name, target_col=None):
        """Perform bivariate analysis on the dataset"""
        print(f"\n=== Bivariate Analysis for {dataset_name} ===")
        
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        
        # If target column is specified, plot relationships with target
        if target_col and target_col in df.columns:
            # Correlation with target
            correlations = df[numeric_cols].corrwith(df[target_col]).sort_values(ascending=False)
            print(f"Correlations with {target_col}:")
            print(correlations)
            
            # Plot relationships with target
            n_cols = 3
            features_to_plot = numeric_cols.drop(target_col)[:6]  # Top 6 features
            n_rows = int(np.ceil(len(features_to_plot) / n_cols))
            
            fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5*n_rows))
            axes = axes.flatten()
            
            for i, col in enumerate(features_to_plot):
                if i < len(axes):
                    axes[i].scatter(df[col], df[target_col], alpha=0.6)
                    axes[i].set_xlabel(col)
                    axes[i].set_ylabel(target_col)
                    axes[i].set_title(f'{col} vs {target_col}')
            
            # Hide any empty subplots
            for j in range(i+1, len(axes)):
                fig.delaxes(axes[j])
            
            plt.tight_layout()
            plt.suptitle(f'Bivariate Analysis - {dataset_name}', y=1.02, fontsize=16)
            plt.show()
        
        # Correlation matrix heatmap
        plt.figure(figsize=(12, 10))
        correlation_matrix = df[numeric_cols].corr()
        mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
        sns.heatmap(correlation_matrix, mask=mask, annot=True, cmap='coolwarm', center=0, 
                   square=True, linewidths=0.5, cbar_kws={"shrink": .8})
        plt.title(f'Correlation Matrix - {dataset_name}', fontsize=16)
        plt.tight_layout()
        plt.show()
    
    def perform_multivariate_analysis(self, df, dataset_name, target_col=None):
        """Perform multivariate analysis on the dataset"""
        print(f"\n=== Multivariate Analysis for {dataset_name} ===")
        
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        
        # PCA for dimensionality reduction
        if len(numeric_cols) > 2:
            # Standardize the data
            scaler = StandardScaler()
            X_scaled = scaler.fit_transform(df[numeric_cols].dropna())
            
            # Perform PCA
            pca = PCA(n_components=2)
            principal_components = pca.fit_transform(X_scaled)
            
            # Create a DataFrame with the principal components
            pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
            
            # Add target variable if available
            if target_col and target_col in df.columns:
                pca_df[target_col] = df[target_col].values
            
            # Plot PCA results
            plt.figure(figsize=(10, 8))
            if target_col and target_col in df.columns:
                scatter = plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df[target_col], 
                                     cmap='viridis', alpha=0.7)
                plt.colorbar(scatter, label=target_col)
            else:
                plt.scatter(pca_df['PC1'], pca_df['PC2'], alpha=0.7)
            
            plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
            plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
            plt.title(f'PCA - {dataset_name}', fontsize=16)
            plt.grid(True, alpha=0.3)
            plt.show()
            
            print(f"PCA Explained Variance Ratio: {pca.explained_variance_ratio_}")
            print(f"Total Variance Explained: {sum(pca.explained_variance_ratio_):.2%}")
        
        # Pairplot for multivariate relationships
        if len(numeric_cols) > 1:
            # Select a subset of columns for pairplot to avoid overcrowding
            cols_for_pairplot = numeric_cols[:5]  # First 5 numeric columns
            if target_col and target_col in df.columns:
                cols_for_pairplot = list(cols_for_pairplot) + [target_col]
            
            pairplot_df = df[cols_for_pairplot].dropna()
            
            if len(pairplot_df) > 0:
                g = sns.pairplot(pairplot_df, hue=target_col if target_col in pairplot_df.columns else None,
                                diag_kind='hist', palette='viridis', plot_kws={'alpha': 0.6})
                g.fig.suptitle(f'Pairplot - {dataset_name}', y=1.02, fontsize=16)
                plt.show()
    
    def feature_engineering(self, df, dataset_name):
        """Perform feature engineering on the dataset"""
        print(f"\n=== Feature Engineering for {dataset_name} ===")
        
        df_engineered = df.copy()
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        
        # Create interaction features
        for i, col1 in enumerate(numeric_cols):
            for j, col2 in enumerate(numeric_cols):
                if i < j:  # Avoid duplicate interactions
                    df_engineered[f'{col1}_x_{col2}'] = df[col1] * df[col2]
        
        # Create polynomial features
        for col in numeric_cols:
            df_engineered[f'{col}_squared'] = df[col] ** 2
            df_engineered[f'{col}_sqrt'] = np.sqrt(np.abs(df[col]))
        
        # Create ratio features
        for i, col1 in enumerate(numeric_cols):
            for j, col2 in enumerate(numeric_cols):
                if i != j and np.all(df[col2] != 0):  # Avoid division by zero
                    df_engineered[f'{col1}_div_{col2}'] = df[col1] / df[col2]
        
        print(f"Original features: {len(df.columns)}")
        print(f"After feature engineering: {len(df_engineered.columns)}")
        
        return df_engineered
    
    def feature_selection(self, df, target_col, problem_type='regression'):
        """Perform feature selection based on target variable"""
        print(f"\n=== Feature Selection for target: {target_col} ===")
        
        # Separate features and target
        X = df.drop(columns=[target_col], errors='ignore').select_dtypes(include=[np.number])
        y = df[target_col]
        
        # Remove columns with too many missing values
        X = X.dropna(axis=1, how='any')
        
        if len(X.columns) == 0:
            print("No features available for selection after removing missing values.")
            return []
        
        # Calculate feature importance
        if problem_type == 'regression':
            # Mutual information for regression
            mi = mutual_info_regression(X, y, random_state=42)
            feature_importance = pd.DataFrame({
                'feature': X.columns,
                'importance': mi
            }).sort_values('importance', ascending=False)
            
            # Random Forest for regression
            rf = RandomForestRegressor(n_estimators=100, random_state=42)
            rf.fit(X, y)
            rf_importance = pd.DataFrame({
                'feature': X.columns,
                'importance': rf.feature_importances_
            }).sort_values('importance', ascending=False)
            
        else:  # classification
            # Mutual information for classification
            mi = mutual_info_classif(X, y, random_state=42)
            feature_importance = pd.DataFrame({
                'feature': X.columns,
                'importance': mi
            }).sort_values('importance', ascending=False)
            
            # Random Forest for classification
            rf = RandomForestClassifier(n_estimators=100, random_state=42)
            rf.fit(X, y)
            rf_importance = pd.DataFrame({
                'feature': X.columns,
                'importance': rf.feature_importances_
            }).sort_values('importance', ascending=False)
        
        # Combine results
        combined_importance = pd.merge(
            feature_importance, rf_importance, 
            on='feature', 
            suffixes=('_mi', '_rf')
        )
        combined_importance['combined_score'] = (
            combined_importance['importance_mi'] + combined_importance['importance_rf']
        ) / 2
        combined_importance = combined_importance.sort_values('combined_score', ascending=False)
        
        print("Top 10 most important features:")
        print(combined_importance.head(10))
        
        # Plot feature importance
        plt.figure(figsize=(12, 8))
        top_features = combined_importance.head(15)
        y_pos = np.arange(len(top_features))
        
        plt.barh(y_pos, top_features['combined_score'], align='center', alpha=0.8)
        plt.yticks(y_pos, top_features['feature'])
        plt.xlabel('Importance Score')
        plt.title('Feature Importance')
        plt.gca().invert_yaxis()
        plt.tight_layout()
        plt.show()
        
        return combined_importance
    
    def prepare_for_ml(self, df, target_col, scale_features=True):
        """Prepare data for machine learning"""
        print(f"\n=== Preparing Data for Machine Learning ===")
        
        # Separate features and target
        X = df.drop(columns=[target_col], errors='ignore')
        y = df[target_col]
        
        # Handle categorical variables
        categorical_cols = X.select_dtypes(exclude=[np.number]).columns
        if len(categorical_cols) > 0:
            X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
        
        # Handle missing values
        X = self.handle_missing_values(X, strategy='knn')
        
        # Scale features if requested
        if scale_features:
            scaler = StandardScaler()
            X_scaled = scaler.fit_transform(X.select_dtypes(include=[np.number]))
            X[X.select_dtypes(include=[np.number]).columns] = X_scaled
        
        print(f"Final dataset shape: {X.shape}")
        print(f"Target variable distribution:\n{y.value_counts() if len(y.unique()) < 10 else y.describe()}")
        
        return X, y
    
    def comprehensive_eda(self):
        """Perform comprehensive EDA on both datasets"""
        print("Starting Comprehensive Exploratory Data Analysis")
        print("="*60)
        
        # Generate realistic data
        self.generate_realistic_sabah_data(nt=50, nx=50, ny=50)
        self.generate_realistic_abudhabi_data(nt=50, nx=50, ny=50)
        
        # Extract features from wells
        sabah_features = self.extract_features_from_wells('sabah')
        abudhabi_features = self.extract_features_from_wells('abudhabi')
        
        # Handle missing values
        sabah_features = self.handle_missing_values(sabah_features)
        abudhabi_features = self.handle_missing_values(abudhabi_features)
        
        # Perform EDA for Sabah dataset
        print("\n" + "="*60)
        print("SABAH DATASET ANALYSIS")
        print("="*60)
        
        self.perform_univariate_analysis(sabah_features, "Sabah Dataset")
        self.perform_bivariate_analysis(sabah_features, "Sabah Dataset", target_col='facies')
        self.perform_multivariate_analysis(sabah_features, "Sabah Dataset", target_col='facies')
        
        # Feature engineering for Sabah
        sabah_engineered = self.feature_engineering(sabah_features, "Sabah Dataset")
        
        # Feature selection for Sabah
        self.feature_selection(sabah_engineered, 'facies', problem_type='classification')
        self.feature_selection(sabah_engineered, 'permeability', problem_type='regression')
        
        # Prepare for ML for Sabah
        X_sabah, y_sabah = self.prepare_for_ml(sabah_engineered, 'facies')
        
        # Perform EDA for Abu Dhabi dataset
        print("\n" + "="*60)
        print("ABU DHABI DATASET ANALYSIS")
        print("="*60)
        
        self.perform_univariate_analysis(abudhabi_features, "Abu Dhabi Dataset")
        self.perform_bivariate_analysis(abudhabi_features, "Abu Dhabi Dataset", target_col='sweet_spot')
        self.perform_multivariate_analysis(abudhabi_features, "Abu Dhabi Dataset", target_col='sweet_spot')
        
        # Feature engineering for Abu Dhabi
        abudhabi_engineered = self.feature_engineering(abudhabi_features, "Abu Dhabi Dataset")
        
        # Feature selection for Abu Dhabi
        self.feature_selection(abudhabi_engineered, 'sweet_spot', problem_type='classification')
        self.feature_selection(abudhabi_engineered, 'permeability', problem_type='regression')
        
        # Prepare for ML for Abu Dhabi
        X_abudhabi, y_abudhabi = self.prepare_for_ml(abudhabi_engineered, 'sweet_spot')
        
        print("\n" + "="*60)
        print("EDA COMPLETE - DATASETS READY FOR MACHINE LEARNING")
        print("="*60)
        
        return {
            'sabah': {'X': X_sabah, 'y': y_sabah, 'features': sabah_engineered},
            'abudhabi': {'X': X_abudhabi, 'y': y_abudhabi, 'features': abudhabi_engineered}
        }

# Run the comprehensive EDA
if __name__ == "__main__":
    eda = ReservoirEDA()
    ml_datasets = eda.comprehensive_eda()
    
    print("\nSummary of prepared datasets:")
    print(f"Sabah dataset shape: {ml_datasets['sabah']['X'].shape}")
    print(f"Abu Dhabi dataset shape: {ml_datasets['abudhabi']['X'].shape}")
```
