## Petrophysical Examples & Python Implementation
### A. Nominal Data
Definition: Qualitative categories without mathematical meaning.
Petrophysical Use Cases:

Lithology classification

Fluid types (Oil/Water/Gas)

Core facies labels

```python
import pandas as pd

# Example: Lithology column
data = pd.DataFrame({
    'Depth': [3000, 3001, 3002],
    'Lithology': ['Sandstone', 'Shale', 'Carbonate']
})

# Convert to categorical
data['Lithology'] = data['Lithology'].astype('category')
print(data.dtypes)
```

### B. Ordinal Data
Definition: Categories with meaningful order but non-uniform intervals.
Petrophysical Use Cases:

Rock quality rankings (1=Poor, 2=Fair, 3=Good)

Drilling difficulty scales

Core fracture intensity grades

```python
# Example: Rock quality index
rock_quality = pd.Series(
    ['Good', 'Fair', 'Poor', 'Good'],
    dtype=pd.CategoricalDtype(categories=['Poor', 'Fair', 'Good'], ordered=True)
)
print(rock_quality)
```

### C. Interval Data
Definition: Numeric values where zero is arbitrary.
Petrophysical Use Cases:

Temperature (°C/°F)

Calendar time (years)

Sonic travel time (μs/ft)

```python
# Example: Temperature log
temp_data = pd.DataFrame({
    'Depth': [3000, 3010, 3020],
    'Temp_C': [120.5, 121.3, 122.1]  # Zero doesn't mean "no temperature"
})
```

## Business Problems:

#### Thermal maturity modeling

#### Time-lapse reservoir monitoring



### D. Ratio Data
Definition: Numeric values with a true zero point.
Petrophysical Use Cases:

Porosity (0 = no pore space)

Permeability (0 = impermeable)

Water saturation (0% = completely dry)

```python
# Example: Core porosity measurements
core_data = pd.DataFrame({
    'SampleID': ['A1', 'A2', 'A3'],
    'Porosity': [0.18, 0.22, 0.0]  # True zero exists
})

# Statistical operations are meaningful
print(f"Mean porosity: {core_data['Porosity'].mean():.2f}")
```
